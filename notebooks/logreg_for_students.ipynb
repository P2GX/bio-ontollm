## Synthetic binary-classification data

*Written by an AI assistant using Copilot SDK in VS Code (GPT-5.6 Terra); proofread by Leonardo.*

We generate two two-dimensional Gaussian clusters. The goal is to learn a linear decision boundary that separates Class 0 from Class 1.

`X_train` contains the feature values, and `y_train` contains the corresponding binary labels.

```python
np.random.seed(42)

n_samples_per_class = 250

X_class0_raw = np.random.normal(loc=0.0, scale=1.0, size=(n_samples_per_class, 2))
X_class1_raw = np.random.normal(loc=0.0, scale=1.0, size=(n_samples_per_class, 2))
transformation = np.eye(2)
X_class0 = X_class0_raw @ transformation
X_class1 = X_class1_raw @ transformation
X_class1 += np.array([3.3, 0.0])

X_raw = np.vstack((X_class0, X_class1))
y_raw = np.hstack(
    (np.zeros(n_samples_per_class), np.ones(n_samples_per_class))
).astype(int)

data = pd.DataFrame(X_raw, columns=["feature1", "feature2"])
data["label"] = y_raw

X_train = data[["feature1", "feature2"]].values
y_train = data["label"].values

m, n = X_train.shape

print(f"Shape of X_train (m, n): {m}, {n}")
```

```python
plt.scatter(X_train[y_train == 0, 0], X_train[y_train == 0, 1], color='tab:blue', label='Class 0', s=20)
plt.scatter(X_train[y_train == 1, 0], X_train[y_train == 1, 1], color='tab:orange', label='Class 1', s=20)
plt.xlabel('Feature1')
plt.ylabel('Feature2')
plt.legend()
plt.savefig("lr-classes.svg", format="svg", bbox_inches="tight")
```
